```python
import os, re, unicodedata, subprocess
from pathlib import Path
import yt_dlp

MODEL = "small"
LANGUAGE = "es"
link = "https://www.youtube.com/watch?v=345IxGgjF9s"

def sanitize_title(title: str) -> str:
    title = unicodedata.normalize("NFKD", title)
    title = title.encode("ascii", "ignore").decode("ascii")
    title = re.sub(r'[<>:"/\\|?*]', "", title)
    title = re.sub(r"\s+", " ", title).strip()
    return title or "untitled"

ydl_opts = {
    "format": "bestaudio/best",
    "outtmpl": "%(title)s.%(ext)s",
    "postprocessors": [{"key": "FFmpegExtractAudio", "preferredcodec": "opus"}],
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info = ydl.extract_info(link, download=True)

# yt-dlp gives you the real final filepath(s)
download = info["requested_downloads"][0]
audio_path = Path(download["filepath"]).with_suffix(".opus")  # after extract audio

# Make output folder (sanitized) next to the audio
folder = audio_path.parent / sanitize_title(audio_path.stem)
folder.mkdir(exist_ok=True)

subprocess.run([
    "whisper", str(audio_path),
    "--language", LANGUAGE,
    "--model", MODEL,
    "--output_dir", str(folder),
], check=True)
```
